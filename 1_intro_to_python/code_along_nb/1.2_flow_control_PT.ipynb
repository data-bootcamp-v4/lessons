## 2. `if`, `elif`, `else`

Use conditional statements when the program should behave differently depending on a condition.

```python
if condition:
    # runs when condition is True
elif another_condition:
    # checked only if the previous condition was False
else:
    # runs when none of the previous conditions were True
```

**Indentation matters:** indented lines belong to the conditional block.


````python
occupancy_rate = 92

if occupancy_rate >= 95:
    print("Almost full")
elif occupancy_rate >= 80:
    print("High occupancy")
else:
    print("Rooms available")
````

## 4. `for` loops: repeat once per item

A `for` loop is useful when you want to perform the same action for every item in a collection.

```python
for item in collection:
    # code repeated for each item
```

Lists, tuples, sets, strings, and dictionaries can all be iterated over.


````python
guests = ["Amina", "Lucas", "Mei"]

for guest in guests:
    print(guest)
````

Notice what changes on each iteration: `guest` temporarily stores the current item from the list.

The loop variable can also be used to access or update related data.


````python
room_status = {
    "101": "ready",
    "102": "cleaning",
    "103": "inspection"
}

for room in room_status:
    print(room, room_status[room])
````

## 5. `while` loops: repeat while a condition is true

A `while` loop is useful when you **do not know in advance how many repetitions are needed**.

```python
while condition:
    # repeated code
```

The condition must eventually become `False`, otherwise the loop will never end.


````python
attempts_left = 3

while attempts_left > 0:
    print("Check-in attempt")
    attempts_left -= 1
````

### A common pattern: ask until the user decides to stop

The condition can depend on a value entered by the user. Imagine a receptionist adding special requests to a booking.

````python
answer = "yes"

while answer == "yes":
    print("Add another request")
    answer = input("Add another request? (yes/no): ").lower()
````

## 6. `for` or `while`?

A useful rule of thumb:

| Situation | Usually use |
|---|---|
| Repeat once for every item in a collection | `for` |
| Repeat until something changes | `while` |

This is not an absolute rule, but it is a good starting point.
