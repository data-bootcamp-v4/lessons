## 1. Why functions?

A function is a **named, reusable block of code** that performs one task.

Functions help us:

- avoid repeating code;
- split a larger problem into smaller tasks;
- test each task independently;
- make programs easier to read and maintain.

A useful way to connect today's topics is:

```text
Data structures → organize data
Flow control    → control what code executes
Functions       → organize code
```


````python
sessions = ["Python Basics", "SQL Fundamentals", "Data Visualization"]

print(f"The conference currently has {len(sessions)} sessions.")
````

The line above uses built-in functions such as `print()` and `len()`.

We can also create our **own** functions.


## 2. Defining and calling a function

A basic function has this structure:

```python
def function_name():

```

**Defining** a function does not run it.

To execute it, we **call** it using its name followed by parentheses.


````python
def show_welcome_message():
    print("Welcome to the conference!")

show_welcome_message()
````

## 3. Parameters and arguments

A **parameter** is a placeholder written in the function definition.

An **argument** is the actual value passed when the function is called.


````python
def show_session(session_name, capacity):
    print(f"{session_name}: {capacity} seats")

show_session("Python Basics", 40)
show_session("SQL Fundamentals", 30)
````

In:

```python
def show_session(session_name, capacity):
```

`session_name` and `capacity` are **parameters**.

In:

```python
show_session("Python Basics", 40)
```

`"Python Basics"` and `40` are **arguments**.

## 4. `return`: sending data back

`print()` **displays** a value.

`return` **sends a value back** to the code that called the function, so the value can be stored or used later.

A useful mental model is:

```text
arguments → function → returned value
```


````python
def calculate_ticket_revenue(ticket_price, tickets_sold):
    revenue = ticket_price * tickets_sold
    return revenue

total_revenue = calculate_ticket_revenue(35, 80)

print(total_revenue)
````

## 5. A practical rule for moving data into and out of functions

Three simple rules are extremely useful when designing functions:

> **If a function needs data from outside → use a parameter.**  
> **If data is created inside a function and you need it outside → use `return`.**  
> **If you only want to show something to the user → use `print()`.**

This makes the flow of data through the program easier to see.


### Avoid hidden dependencies when possible

Python allows a function to use variables defined outside it:

```python
tax_rate = 0.21

def calculate_tax(price):
    return price * tax_rate
```

This works, but the function now depends on a variable that is defined somewhere else in the program.

A clearer design is usually to pass that value explicitly:


````python
def calculate_tax(price, tax_rate):
    return price * tax_rate

rate = 0.21
tax = calculate_tax(100, rate)

print(tax)
````

Now the function's dependencies are visible in its parameters:

```text
price ────────┐
              ├──► calculate_tax() ───► tax
tax_rate ─────┘
```

A good rule of thumb:

> **A function should receive through its parameters the external data it needs to do its job.**

This makes functions more independent, reusable, easier to understand, and easier to test.

## 6. Local variables and passing results between functions

Variables created inside a function belong to that function.

If another part of the program needs a value calculated inside the function, **return it**.

If another function needs that value, pass it as an **argument**.


````python
def calculate_event_cost(venue_cost, catering_cost):
    total = venue_cost + catering_cost
    return total

def display_event_cost(total):
    print(f"Event cost: €{total}")

event_cost = calculate_event_cost(1200, 800)
display_event_cost(event_cost)
````

## 8. Returning multiple values

A function can return more than one value by separating them with commas.

Python groups those values into a tuple, and we can unpack them into separate variables.


````python
def calculate_attendance_summary(attendees, capacity):
    occupancy = attendees / capacity * 100
    return attendees, occupancy

total_attendees, occupancy_rate = calculate_attendance_summary(72, 90)

print(total_attendees)
print(occupancy_rate)
````

## 9. `print()` vs `return`

Both are useful, but they do different jobs.

### Use `return` when:

- another part of the program needs the result;
- another function needs the result;
- you want to store the result;
- you want to calculate something else with it.

### Use `print()` when:

- the function's job is simply to display information to the user.


````python
def calculate_available_seats(capacity, registered):
    return capacity - registered

def print_available_seats(session_name, available):
    print(f"{session_name}: {available} seats available")

available = calculate_available_seats(40, 31)

print_available_seats("Python Basics", available)
````

## 10. Building a program from small functions

A larger program can often be understood as a sequence of small functions:

```text
get data
   ↓
process data
   ↓
calculate results
   ↓
display results
```

Each function should have a **clear responsibility**.

The returned result of one function can become the argument of another.


````python
def calculate_early_bird_discount(ticket_price):
    if ticket_price >= 50:
        return ticket_price * 0.10

    return 0

def calculate_final_price(ticket_price, discount):
    return ticket_price - discount

def print_ticket_price(final_price):
    print(f"Final ticket price: €{final_price}")

ticket_price = 60

discount = calculate_early_bird_discount(ticket_price)
final_price = calculate_final_price(ticket_price, discount)
print_ticket_price(final_price)
````
