## 1. Why OOP?

So far, we have kept **data** and **functions** separate.

Object-Oriented Programming groups related data and behavior together.

In a logistics company, a package has:

- data: tracking id, destination, weight, status;
- behavior: update status, calculate a shipping cost, display information.

A **class** is the blueprint. An **object** is one concrete instance created from that blueprint.

````python
class Package:
    pass

package_a = Package()
package_b = Package()

print(type(package_a))
print(package_a is package_b)
````

Both objects come from the same class, but they are different objects.

This distinction will become important throughout the rest of the bootcamp: NumPy arrays, Pandas DataFrames and many machine-learning models are also objects created from classes.

## 2. `__init__`, `self`, and instance attributes

`__init__()` runs automatically when a new object is created.

`self` refers to the **current object**.

```python
class MyClass:
    def __init__(self, value):
        self.value = value
```

Each object gets its own copy of its instance attributes.

````python
class Package:
    def __init__(self, tracking_id, destination, weight):
        self.tracking_id = tracking_id
        self.destination = destination
        self.weight = weight
        self.status = "created"

package_a = Package("PKG-001", "Madrid", 2.4)
package_b = Package("PKG-002", "Lisbon", 5.1)

print(package_a.destination)
print(package_b.destination)
print(package_a.status)
````

## 3. Methods: behavior that belongs to an object

A **method** is a function defined inside a class.

It receives `self`, so it can read or modify that object's attributes.

````python
class Package:
    def __init__(self, tracking_id, destination, weight):
        self.tracking_id = tracking_id
        self.destination = destination
        self.weight = weight
        self.status = "created"

    def update_status(self, new_status):
        self.status = new_status

    def get_summary(self):
        return f"{self.tracking_id} → {self.destination} ({self.status})"

package = Package("PKG-003", "Paris", 1.8)
package.update_status("in transit")
print(package.get_summary())
````

## 4. Default values

````python
class Package:
    def __init__(self, tracking_id, destination, weight, insured=False):
        self.tracking_id = tracking_id
        self.destination = destination
        self.weight = weight
        self.insured = insured

standard = Package("PKG-005", "Rome", 3.0)
insured = Package("PKG-006", "Berlin", 2.1, insured=True)

print(standard.insured)
print(insured.insured)
````

## 5. Class attributes: data shared by all instances

An **instance attribute** belongs to one object.

A **class attribute** belongs to the class and is shared across its instances.

A common use is keeping a counter.

````python
class Package:
    package_count = 0

    def __init__(self, destination):
        Package.package_count += 1
        self.package_id = Package.package_count
        self.destination = destination

package_a = Package("Madrid")
package_b = Package("Lisbon")
package_c = Package("Paris")

print(package_a.package_id)
print(package_b.package_id)
print(Package.package_count)
````

## 6. Objects can interact with other objects

Methods can receive other objects as arguments.

This is where OOP becomes especially useful: objects can collaborate while keeping their own state.

````python
class Truck:
    def __init__(self, truck_id):
        self.truck_id = truck_id
        self.packages = []

    def load_package(self, package):
        self.packages.append(package)

    def unload_package(self, package):
        if package in self.packages:
            self.packages.remove(package)

class Package:
    def __init__(self, tracking_id):
        self.tracking_id = tracking_id

package = Package("PKG-007")
truck_a = Truck("TR-A")
truck_b = Truck("TR-B")

truck_a.load_package(package)
truck_a.unload_package(package)
truck_b.load_package(package)

print([p.tracking_id for p in truck_a.packages])
print([p.tracking_id for p in truck_b.packages])
````

## 7. Inheritance: creating a specialized class

Inheritance lets a **child class** reuse attributes and methods from a **parent class**.

```python
class Child(Parent):
    ...
```

A child class can:

- reuse inherited methods;
- add new attributes or methods;
- replace (override) inherited methods.

````python
class Package:
    def __init__(self, tracking_id, destination, weight):
        self.tracking_id = tracking_id
        self.destination = destination
        self.weight = weight
        self.status = "created"

    def update_status(self, new_status):
        self.status = new_status

class ExpressPackage(Package):
    pass

express = ExpressPackage("EXP-001", "Amsterdam", 1.2)
express.update_status("priority transit")
print(express.status)
````

The child class did not define `update_status()`, but it can use it because it inherited that method from `Package`.

## 8. `super()`: initialize the parent part of the object

When a child class has its own `__init__()`, use `super().__init__(...)` to reuse the parent's initialization instead of rewriting it.

````python
class Package:
    def __init__(self, tracking_id, destination, weight):
        self.tracking_id = tracking_id
        self.destination = destination
        self.weight = weight

class ExpressPackage(Package):
    def __init__(self, tracking_id, destination, weight, express_fee=8):
        super().__init__(tracking_id, destination, weight)
        self.express_fee = express_fee

express = ExpressPackage("EXP-002", "Brussels", 2.0)
print(express.destination)
print(express.express_fee)
````

## 9. Overriding methods

A child class can define a method with the same name as one in the parent class.

The child's version then takes priority.

You can still call the parent version with `super().method_name(...)`.

````python
class Package:
    def shipping_cost(self):
        return 5

class ExpressPackage(Package):
    def __init__(self, express_fee=8):
        self.express_fee = express_fee

    def shipping_cost(self):
        return super().shipping_cost() + self.express_fee

standard = Package()
express = ExpressPackage()

print(standard.shipping_cost())
print(express.shipping_cost())
````
