## 1. Why functions?

A function is a **named, reusable block of code** that performs one task.

Functions help us:

- avoid repeating code;
- split a larger problem into smaller tasks;
- test each task independently;
- make programs easier to read and maintain.


````python
sessions = ["Python Basics", "SQL Fundamentals", "Data Visualization"]

print(f"The conference currently has {len(sessions)} sessions.")
````

The line above uses built-in functions such as `print()` and `len()`. We can also create our **own** functions.


## 2. Defining and calling a function

A basic function has this structure:

```python
def function_name():
    # code that belongs to the function
```

Defining a function does **not** run it. To execute it, we **call** it using its name followed by parentheses.


````python
def show_welcome_message():
    print("Welcome to the conference!")

show_welcome_message()
````

## 3. Parameters and arguments

A **parameter** is a placeholder written in the function definition.

An **argument** is the actual value passed when the function is called.


````python
def show_session(session_name, capacity):
    print(f"{session_name}: {capacity} seats")

show_session("Python Basics", 40)
show_session("SQL Fundamentals", 30)
````

In `show_session(session_name, capacity)`:

- `session_name` and `capacity` are **parameters**;
- `"Python Basics"` and `40` are **arguments** in the first call.

## 4. `return`: sending data back

`print()` **displays** a value.

`return` **sends a value back** to the code that called the function, so that value can be stored or used later.


````python
def calculate_ticket_revenue(ticket_price, tickets_sold):
    revenue = ticket_price * tickets_sold
    return revenue

total_revenue = calculate_ticket_revenue(35, 80)
print(total_revenue)
````

A useful mental model is:

```text
arguments → function → returned value
```

If another part of the program needs the result, prefer `return`.

## 5. Local variables and moving data between functions

Variables created inside a function belong to that function. If the rest of the program needs a value calculated inside the function, **return it**.

Likewise, if a function needs data produced elsewhere, **pass it as a parameter**.

This is the key idea for building programs from several functions.


````python
def create_attendee_name():
    attendee_name = "Ana"
    return attendee_name

def greet_attendee(name):
    print(f"Hello, {name}!")

name = create_attendee_name()
greet_attendee(name)
````

## 6. Functions with no parameters

Not every function needs input parameters. A function can obtain or create the data it needs internally.

For example, a function may ask the user for information and return the result.


````python
def get_attendee_name():
    name = input("Attendee name: ")
    return name

attendee_name = get_attendee_name()
print(f"The attendee name is: {attendee_name}")
````

## 7. Returning multiple values

A function can return more than one value by separating them with commas. Python groups those values into a tuple, and we can unpack them into separate variables.


````python
def calculate_attendance_summary(attendees, capacity):
    occupancy = attendees / capacity * 100
    return attendees, occupancy

total_attendees, occupancy_rate = calculate_attendance_summary(72, 90)

print(total_attendees)
print(occupancy_rate)
````

## 8. Functions that print vs functions that return

Both styles are valid, but they serve different purposes.

**Return a value** when:
- another function needs it;
- you want to store it;
- you want to calculate something else with it.

**Print a value** when the function's job is simply to display information to the user.


````python
def calculate_available_seats(capacity, registered):
    return capacity - registered

def print_available_seats(session_name, available):
    print(f"{session_name}: {available} seats available")

available = calculate_available_seats(40, 31)
print_available_seats("Python Basics", available)
````

## 9. Building a program from small functions

A larger program is often just a sequence of small functions:

```text
get data
   ↓
process data
   ↓
calculate results
   ↓
display results
```

Each function **should have a clear responsibility**.


````python
def calculate_early_bird_discount(ticket_price):
    if ticket_price >= 50:
        return ticket_price * 0.10
    return 0

def calculate_final_price(ticket_price, discount):
    return ticket_price - discount

def print_ticket_price(final_price):
    print(f"Final ticket price: €{final_price}")

ticket_price = 60
discount = calculate_early_bird_discount(ticket_price)
final_price = calculate_final_price(ticket_price, discount)
print_ticket_price(final_price)
````
