## 4. `__init__()`: initialize a new object

An empty class is not very useful yet.

Usually, every object needs to remember some information.

`__init__()` runs automatically when a new object is created:


````python
class Package:
    def __init__(self, tracking_id, destination, weight):
        self.tracking_id = tracking_id
        self.destination = destination
        self.weight = weight


package_a = Package("PKG-001", "Madrid", 2.4)
package_b = Package("PKG-002", "Lisbon", 5.1)

print(package_a.destination)
print(package_b.destination)
````

### Parameters vs attributes

Look carefully at this line:

```python
self.destination = destination
```

The two names have different roles:

```text
destination        parameter received by __init__()
     │
     ▼
self.destination   attribute stored inside the object
```

The parameter provides the value.

The attribute remembers that value as part of the object's state.


## 5. What exactly is `self`?

`self` refers to the **current object**.

When we create:


````python
package_a = Package("PKG-001", "Madrid", 2.4)
package_b = Package("PKG-002", "Lisbon", 5.1)
````

the same class creates two independent objects:

```text
package_a
├── tracking_id → "PKG-001"
├── destination → "Madrid"
└── weight      → 2.4


package_b
├── tracking_id → "PKG-002"
├── destination → "Lisbon"
└── weight      → 5.1
```

Inside the class, `self.destination` means:

> the `destination` attribute of **this particular object**.

That is why different objects can store different values in attributes with the same name.

## 6. Methods: behavior that belongs to an object

A **method** is a function defined inside a class.

Because an instance method receives `self`, it can read or modify the state of the current object.


````python
class Package:
    def __init__(self, tracking_id, destination, weight):
        self.tracking_id = tracking_id
        self.destination = destination
        self.weight = weight
        self.status = "created"

    def update_status(self, new_status):
        self.status = new_status

    def get_summary(self):
        return f"{self.tracking_id} → {self.destination} ({self.status})"


package = Package("PKG-003", "Paris", 1.8)

package.update_status("in transit")

print(package.status)
print(package.get_summary())
````

We can think of an object as combining:

```text
OBJECT
│
├── attributes → data / state
│
└── methods    → behavior / actions
```

This also connects with Tuesday's lesson on functions:

```text
FUNCTION                         METHOD

def calculate(x):               class Package:
    ...                             def update_status(self, status):
                                       ...

calculate(x)                    package.update_status(status)
```

A method is still a function, but it belongs to a class and can work with an object's state through `self`.

## 8. Instance attributes vs class attributes

So far, our attributes have belonged to individual objects.

These are **instance attributes**.

Sometimes we need information that belongs to the class itself and is shared across all its objects.

That is a **class attribute**.


````python
class Truck:
    truck_count = 0

    def __init__(self):
        Truck.truck_count += 1
        self.truck_id = "truck_" + str(Truck.truck_count)


truck_a = Truck()
truck_b = Truck()
truck_c = Truck()

print(Truck.truck_count)

print(truck_a.truck_id)
print(truck_b.truck_id)
print(truck_c.truck_id)
````

The two kinds of attributes now have clearly different roles:

```text
CLASS ATTRIBUTE
Truck.truck_count
        ↓
shared by the class
        ↓
counts how many Truck objects have been created


INSTANCE ATTRIBUTE
truck_a.truck_id → "truck_1"
truck_b.truck_id → "truck_2"
truck_c.truck_id → "truck_3"
        ↓
belongs to each individual object
```

Notice the sequence:

```text
Truck.truck_count starts at 0

create truck_a
    Truck.truck_count → 1
    truck_a.truck_id  → "truck_1"

create truck_b
    Truck.truck_count → 2
    truck_b.truck_id  → "truck_2"

create truck_c
    Truck.truck_count → 3
    truck_c.truck_id  → "truck_3"
```

The class attribute keeps the **shared counter**.

Each object uses the current counter value to build and store its **own identifier**.


## 9. Objects can interact with other objects

An object does not have to work alone.

Methods and functions can receive other objects as arguments.


````python
class Package:
    def __init__(self, tracking_id):
        self.tracking_id = tracking_id


class Truck:
    def __init__(self, truck_id):
        self.truck_id = truck_id
        self.packages = []

    def load_package(self, package):
        self.packages.append(package)

    def unload_package(self, package):
        if package in self.packages:
            self.packages.remove(package)


package = Package("PKG-007")
truck = Truck("TR-A")

truck.load_package(package)

print(truck.packages)
print(truck.packages[0].tracking_id)
````

Notice what is stored inside `truck.packages`:

```text
Truck object
    │
    └── packages
           │
           └── Package object
```

The list does not need to contain only strings or numbers. It can contain objects too.

## 10. Inheritance: create a specialized class

Sometimes two classes represent closely related concepts.

An **ExpressPackage** is still a **Package**, but it may need some additional behavior.

Inheritance lets a **child class** reuse attributes and methods from a **parent class**.

```python
class Child(Parent):
    ...
```


````python
class Package:
    def __init__(self, tracking_id, destination, weight):
        self.tracking_id = tracking_id
        self.destination = destination
        self.weight = weight
        self.status = "created"

    def update_status(self, new_status):
        self.status = new_status


class ExpressPackage(Package):
    pass


express = ExpressPackage("EXP-001", "Amsterdam", 1.2)

express.update_status("priority transit")

print(express.destination)
print(express.status)
````

`ExpressPackage` does not define `__init__()` or `update_status()`.

It can use them because they are inherited from `Package`.

```text
          Package
          PARENT
             │
             │ inheritance
             ▼
      ExpressPackage
           CHILD
```

A child class can:

- reuse inherited behavior;
- add new attributes or methods;
- replace inherited behavior.

## 11. `super()`: reuse parent initialization

Suppose the child class needs an additional attribute.

We could rewrite all the initialization code, but that would duplicate the parent logic.

Instead, we can use:

```python
super().__init__(...)
```

to run the parent constructor.


````python
class Package:
    def __init__(self, tracking_id, destination, weight):
        self.tracking_id = tracking_id
        self.destination = destination
        self.weight = weight


class ExpressPackage(Package):
    def __init__(self, tracking_id, destination, weight, express_fee=8):
        super().__init__(tracking_id, destination, weight)

        self.express_fee = express_fee


express = ExpressPackage(
    "EXP-002",
    "Brussels",
    2.0
)

print(express.tracking_id)
print(express.destination)
print(express.weight)
print(express.express_fee)
````

Conceptually:

```text
ExpressPackage.__init__()
          │
          ├── super().__init__(...)
          │          ↓
          │    initialize Package part
          │
          └── self.express_fee = ...
                     ↓
              initialize specialized part
```

`super()` lets us reuse the initialization already defined by the parent instead of copying it.

## 12. Overriding methods

A child class can define a method with the **same name** as a method in the parent class.

This is called **method overriding**.

The child's implementation takes priority for objects of the child class.


````python
class Package:
    def shipping_cost(self):
        return 5


class ExpressPackage(Package):
    def __init__(self, express_fee=8):
        self.express_fee = express_fee

    def shipping_cost(self):
        return 5 + self.express_fee


standard = Package()
express = ExpressPackage()

print(standard.shipping_cost())
print(express.shipping_cost())
````

Compare the two situations:

```text
INHERIT
child does not define the method
        ↓
use parent's implementation


OVERRIDE
child defines a method with the same name
        ↓
use child's implementation
```


## 13. Reusing the parent method with `super()`

Sometimes we do not want to completely replace the parent behavior.

We want to **extend** it.

Then we can call the parent's method with `super()`:


````python
class Package:
    def shipping_cost(self):
        return 5


class ExpressPackage(Package):
    def __init__(self, express_fee=8):
        self.express_fee = express_fee

    def shipping_cost(self):
        normal_cost = super().shipping_cost()

        return normal_cost + self.express_fee


express = ExpressPackage()

print(express.shipping_cost())
````

The child method now says, conceptually:

```text
get the normal Package shipping cost
                ↓
       super().shipping_cost()
                ↓
          add express fee
                ↓
          return new cost
```

So `super()` is not only for `__init__()`.

It can call other parent methods too.
