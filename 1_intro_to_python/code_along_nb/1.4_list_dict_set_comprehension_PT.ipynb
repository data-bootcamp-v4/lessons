## 1. Why comprehensions?

A comprehension is a compact way to create a new collection from an iterable.

Compare these two pieces of code:

````python
courses = ["python", "sql", "tableau", "statistics"]

upper_courses = []
for course in courses:
    upper_courses.append(course.upper())

print(upper_courses)
````

The same transformation can be written as a list comprehension:

````python
upper_courses = [course.upper() for course in courses]
print(upper_courses)
````

### The basic pattern

```python
[expression for item in iterable]
```

Read it from left to right as:

> "Create a new list using this **expression** for every **item** in this **iterable**."

## 2. The three comprehension patterns

The same three patterns appear in list, set and dictionary comprehensions.

### Pattern A — Transform every element

```python
[expression for item in iterable]
```

````python
scores = [72, 58, 91, 64]

curved_scores = [score + 5 for score in scores]
curved_scores
````

### ⚡ Quick check

### Pattern B — Filter with `if`

Use this pattern when some elements should be **excluded**.

```python
[expression for item in iterable if condition]
```

````python
scores = [72, 48, 91, 55, 84]

passing_scores = [score for score in scores if score >= 60]
passing_scores
````

### ⚡ Quick check

````python
courses = ["sql", "python", "tableau", "excel", "statistics"]

long_names = [course for course in courses if ____________________]
long_names
````

### Pattern C — Transform conditionally with `if/else`

Use this pattern when **every element remains**, but its output depends on a condition.

```python
[expression_if_true if condition else expression_if_false for item in iterable]
```

Notice that `if/else` comes **before** the `for` because it chooses between two output expressions.

````python
scores = [72, 48, 91, 55]

labels = ["pass" if score >= 60 else "review" for score in scores]
labels
````

## 3. Dictionary comprehensions

A dictionary comprehension follows the same idea, but a dictionary needs **two expressions**:

```python
{key_expression: value_expression for item in iterable}
```

For every iteration, Python must decide:

1. **What will the key be?**
2. **What will the value be?**

Both expressions may transform the original item.

````python
courses = ["python", "sql", "tableau"]

name_lengths = {course: len(course) for course in courses}
name_lengths
````

### Dictionary pattern A — No condition

```python
{key_expression: value_expression for item in iterable}
```

````python
students = ["Ana", "Bruno", "Carla"]
scores = [82, 67, 94]

# One common real-world use is building a dictionary from paired values.
grades = {student: score for student, score in zip(students, scores)}
grades
````

> `zip()` is useful when two iterables belong together. You do not need to master it for the Lab, but it is a useful extension to know.

### Dictionary pattern B — Filter with `if`

Use a trailing `if` when you want to **remove complete key-value pairs**.

```python
{key_expression: value_expression for item in iterable if condition}
```

````python
grades = {"Ana": 82, "Bruno": 48, "Carla": 94, "Diego": 55}

passing_grades = {
    student: score
    for student, score in grades.items()
    if score >= 60
}

passing_grades
````

### Dictionary pattern C — Conditional key/value expressions

A dictionary comprehension can also use `if/else` **inside the key expression, the value expression, or both**.

```python
{
    key_if_true if condition else key_if_false:
    value_if_true if condition else value_if_false
    for item in iterable
}
```

This does **not filter items out**. Every iteration still produces one key-value pair; only the key and/or value changes.

````python
grades = {"Ana": 82, "Bruno": 48, "Carla": 94}

grade_labels = {
    student.upper(): "pass" if score >= 60 else "review"
    for student, score in grades.items()
}

grade_labels
````

> **Key idea:** a trailing `if` decides **whether a pair exists**. An `if/else` inside the key/value expressions decides **what that pair looks like**.

## 4. Set comprehensions

Set comprehensions use the same three patterns as list comprehensions, but return unique values.

```python
{expression for item in iterable}
```

````python
enrollments = ["python", "sql", "python", "tableau", "sql"]

unique_courses = {course for course in enrollments}
unique_courses
````

### ⚡ Quick check

Create a set containing the uppercase version of each unique course name.


````python
enrollments = ["python", "sql", "python", "tableau", "sql"]

unique_upper = {__________________ for course in enrollments}
unique_upper
````

Set comprehensions can also filter or use conditional expressions:

````python
courses = ["sql", "python", "tableau", "excel"]

long_names = {course for course in courses if len(course) > 5}
labels = {"long" if len(course) > 5 else "short" for course in courses}

print(long_names)
print(labels)
````

## 6. Choosing between a loop and a comprehension

Comprehensions are useful when you are **creating a collection from another iterable**.

Prefer a regular loop when:

- the logic needs several steps;
- you need multiple side effects such as several `print()` calls;
- the comprehension becomes difficult to read.

Shorter code is not automatically better code.
