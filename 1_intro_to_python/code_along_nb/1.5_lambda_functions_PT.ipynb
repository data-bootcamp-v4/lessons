## 1. Lambda functions

A `lambda` is a **small anonymous function** written in a single expression.

```python
lambda arguments: expression
```

It is most useful when a function is needed **briefly**, often as an argument to another function.

### From `def` to `lambda`

These two functions do the same thing:

````python
def apply_discount(price):
    return price * 0.9

apply_discount_lambda = lambda price: price * 0.9

print(apply_discount(80))
print(apply_discount_lambda(80))
````

## 2. Using `lambda` with `sorted()`

A common real-world use of lambda is telling `sorted()` **which value should control the ordering**.

```python
sorted(iterable, key=function, reverse=False)
```

The `key` function receives each element and returns the value Python should use for sorting.

````python
destinations = [
    ("Lisbon", 120),
    ("Rome", 150),
    ("Berlin", 95),
    ("Athens", 135)
]

# Sort by estimated daily cost
sorted_destinations = sorted(destinations, key=lambda destination: destination[1])
sorted_destinations
````

## 3. `map()` — transform every element

`map()` applies a function to **every element** of an iterable.

```python
map(function, iterable)
```

It returns a `map` object, so we often convert it to a list to inspect the results.

````python
temperatures_c = [12, 18, 25, 31]

temperatures_f = list(map(lambda c: c * 9/5 + 32, temperatures_c))
temperatures_f
````

Think of `map()` as:

> **same number of elements, transformed values**

This is conceptually similar to a list comprehension used only for transformation.

## 4. `filter()` — keep matching elements

`filter()` keeps the elements for which a function returns `True`.

```python
filter(function, iterable)
```

It also returns an iterator, so we normally wrap it with `list()` when we want a list.

````python
ratings = [4.8, 3.9, 4.6, 2.8, 4.2]

high_ratings = list(filter(lambda rating: rating >= 4.5, ratings))
high_ratings
````

Think of `filter()` as:

> **same type of elements, but only some remain**

This is conceptually similar to a comprehension with a final `if`.

## 5. `reduce()` — combine many values into one

Unlike `map()` and `filter()`, `reduce()` usually turns an iterable into **one accumulated result**.

It lives in `functools`:

```python
from functools import reduce
reduce(function, iterable)
```

The function receives two values:

1. the accumulated result so far;
2. the next element.

````python
from functools import reduce

daily_visitors = [120, 95, 135, 150]

total_visitors = reduce(lambda total, visitors: total + visitors, daily_visitors)
total_visitors
````

### What happens step by step?

For `[120, 95, 135, 150]`:

```text
120 + 95  -> 215
215 + 135 -> 350
350 + 150 -> 500
```

So `reduce()` returns `500`.

## 6. Putting the patterns together

Use this mental model:

| Tool | Question | Typical result |
|---|---|---|
| `map()` | What should happen to **every** element? | iterable → iterable |
| `filter()` | Which elements should I **keep**? | iterable → smaller iterable |
| `reduce()` | How do I **combine** everything? | iterable → one value |
| `sorted(..., key=lambda ...)` | Which part of each element controls the **order**? | iterable → sorted list |

`lambda` is often the short function supplied to these tools.
