## 1. Why comprehensions?

Yesterday we learned how loops allow us to repeat operations.

A very common pattern is:

1. start with an iterable;
2. process each item;
3. store the results in a new collection.

For example:


````python
cities = ["madrid", "berlin", "paris", "lisbon"]

upper_cities = []

for city in cities:
    upper_cities.append(city.upper())

print(upper_cities)
````

Python provides a compact syntax for this pattern:


````python
upper_cities = [city.upper() for city in cities]

print(upper_cities)
````

Both versions produce the same result.

A useful mental model is:

```text
iterable
   ↓
process each item
   ↓
new collection
```

> A comprehension is a compact way to **create a new collection from an iterable**.

It is not simply a way to make every `for` loop shorter.

# 2. Three fundamental comprehension patterns

Most comprehensions that we need today follow one of three patterns.

We will first learn them with **lists**, and then reuse the same ideas with dictionaries and sets.


## Pattern A — Transform every item

```python
[expression for item in iterable]
```

The expression tells Python **what each item becomes**.


````python
prices = [10, 25, 40, 60]

prices_with_tax = [price * 1.21 for price in prices]

print(prices_with_tax)
````

## Pattern B — Filter with `if`

Sometimes we only want some elements to enter the new collection.

```python
[expression for item in iterable if condition]
```

The final `if` answers:

> **Should this item enter the new collection?**


````python
scores = [45, 72, 88, 51, 93, 67]

passing_scores = [score for score in scores if score >= 60]

print(passing_scores)
````

## Pattern C — Transform conditionally with `if/else`

Sometimes every element should remain in the new collection, but we want to transform it differently depending on a condition.

```python
[value_if_true if condition else value_if_false for item in iterable]
```

Here the condition answers:

> **What should this item become?**

Compare the two ideas:

```text
expression for item in iterable if condition
                               ↑
                             FILTER


value_a if condition else value_b for item in iterable
        ↑
   CHOOSE THE OUTPUT
```


````python
scores = [45, 72, 88, 51, 93, 67]

results = [
    "Pass" if score >= 60 else "Fail"
    for score in scores
]

print(results)
````

# 3. Dictionary comprehensions

A dictionary comprehension creates **key-value pairs**:

```python
{key_expression: value_expression for item in iterable}
```

This is an important difference from a list comprehension.

For every iteration, we must decide:

```text
What will the KEY be?
What will the VALUE be?
```

Both the key and the value are **expressions**.


````python
cities = ["Madrid", "Berlin", "Paris"]

name_lengths = {
    city: len(city)
    for city in cities
}

print(name_lengths)
````

## 3.1 Transforming dictionary values

When iterating through an existing dictionary, `.items()` gives us both the key and the value.


````python
temperatures_c = {
    "Madrid": 24,
    "Berlin": 16,
    "Paris": 19
}

temperatures_f = {
    city: temperature * 9 / 5 + 32
    for city, temperature in temperatures_c.items()
}

print(temperatures_f)
````

## 3.2 Filtering dictionary pairs

A trailing `if` decides whether the **entire key-value pair** enters the new dictionary:

```python
{
    key_expression: value_expression
    for key, value in dictionary.items()
    if condition
}
```


````python
scores = {
    "Ana": 82,
    "Bruno": 55,
    "Carla": 91,
    "David": 48
}

passing_scores = {
    student: score
    for student, score in scores.items()
    if score >= 60
}

print(passing_scores)
````

## 3.3 Conditional expression in the VALUE

Because the value is an expression, it can contain `if/else`.

Every key-value pair remains in the dictionary, but the **value changes depending on the condition**.


````python
scores = {
    "Ana": 82,
    "Bruno": 55,
    "Carla": 91
}

status = {
    student: ("Pass" if score >= 60 else "Fail")
    for student, score in scores.items()
}

print(status)
````

## 3.4 Conditional expression in the KEY

The **key is also an expression**.

Therefore, conditional logic can be applied to the key too:


````python
scores = {
    "Ana": 82,
    "Bruno": 55,
    "Carla": 91
}

labelled_scores = {
    ("PASS_" + student if score >= 60 else "FAIL_" + student): score
    for student, score in scores.items()
}

print(labelled_scores)
````

This gives us an important general rule:

> **In a dictionary comprehension, both the key and the value are expressions. Either one can therefore contain conditional logic.**

So we have three independent possibilities:

```text
trailing if
    → decides WHETHER the key-value pair exists

if/else in key_expression
    → decides WHAT THE KEY becomes

if/else in value_expression
    → decides WHAT THE VALUE becomes
```


## 3.5 Combining the three ideas

These mechanisms can also be combined.

Read this example from the inside out:


````python
scores = {
    "Ana": 82,
    "Bruno": 65,
    "Carla": 91,
    "David": 48
}

result = {
    ("HIGH_" + student if score >= 80 else "MEDIUM_" + student):
    ("Excellent" if score >= 90 else "Approved")
    for student, score in scores.items()
    if score >= 60
}

print(result)
````

Here:

1. `if score >= 60` decides **whether the pair enters the dictionary**;
2. the `if/else` in the key decides **which key is created**;
3. the `if/else` in the value decides **which value is created**.

You do **not** need to write complicated comprehensions like this all the time.

The purpose of this example is to understand where each condition acts.

# 4. Set comprehensions

Set comprehensions use the same basic patterns:

```python
{expression for item in iterable}
```

Remember: sets contain **unique values**.


````python
departments = ["Data", "Web", "Data", "UX", "Web"]

unique_departments = {
    department.upper()
    for department in departments
}

print(unique_departments)
````

Filtering works exactly as before:


````python
numbers = [2, 3, 4, 5, 6, 7, 8]

even_squares = {
    number ** 2
    for number in numbers
    if number % 2 == 0
}

print(even_squares)
````

# 6. Loop or comprehension?

Comprehensions are useful, but they are not replacements for every loop.

### A comprehension is a good choice when:

- you are creating a new collection from an iterable;
- the transformation is short;
- the filtering rule is easy to understand;
- the result remains readable.

### Prefer a regular loop when:

- several different operations happen in each iteration;
- the logic has many steps;
- there are several side effects;
- the comprehension becomes difficult to read.

> **Shorter code is not automatically better code.**


````python
# Clear comprehension
squares = [number ** 2 for number in range(10)]

# A longer loop may be clearer when several things must happen
for number in range(10):
    square = number ** 2
    print(f"Processing {number}")
    print(f"Square: {square}")
````

## 7. Creating dictionaries with `zip()`

`zip()` allows us to iterate through two or more iterables at the same time.


````python
names = ["Ana", "Bruno", "Carla"]
scores = [82, 67, 94]

grades = {
    student: score
    for student, score in zip(names, scores)
}

print(grades)
````

## 8. Nested comprehensions

Comprehensions can contain more than one `for`.

For example, a nested list can be flattened:


````python
matrix = [
    [1, 2],
    [3, 4],
    [5, 6]
]

flat = [
    number
    for row in matrix
    for number in row
]

print(flat)
````

The order follows the equivalent nested loop:


````python
flat = []

for row in matrix:
    for number in row:
        flat.append(number)

print(flat)
````

Nested comprehensions are valid Python, but readability matters.

If you need to stop and decode the comprehension, a normal loop may be the better choice.
