## 1. Basic data types
Python stores different kinds of values using different **data types**. The four types we will use most today are:

| Type | Example | Typical use |
|---|---|---|
| `int` | `42` | whole numbers |
| `float` | `19.95` | decimal numbers |
| `str` | `"Madrid"` | text |
| `bool` | `True` | yes/no conditions |

Use `type()` when you are unsure what kind of value you have.

````python
campaign_days = 28
cost_per_click = 1.95
campaign_name = "Summer Launch"
is_active = True

print(type(campaign_days))
print(type(cost_per_click))
print(type(campaign_name))
print(type(is_active))
````

## 4. Data structures
Data structures let us store **collections of values**.

Today we focus mainly on **lists** and **dictionaries** because they appear constantly in Python data work.

## 4.1 Lists
A **list** is an ordered, mutable collection.

```python
[ ]
```

Lists are useful when you have several values that belong together and their **position/order matters**.

````python
channels = ["email", "social", "search"]
print(channels)
````

### Accessing values: indexing
Python starts counting at **0**. Negative indexes count from the end.

````python
channels = ["email", "social", "search"]

print(channels[0])   # first
print(channels[1])   # second
print(channels[-1])  # last
````

### Slicing
A slice selects part of a sequence: `start:end`. The `end` position is **not included**.

````python
weekly_leads = [120, 180, 95, 210, 160]
print(weekly_leads[1:4])
````

### Modifying a list
Lists are **mutable**, so we can change them after creation.

````python
audiences = ["new_users", "returning_users", "newsletter"]

audiences.append("vip_customers")
audiences[1] = "repeat_buyers"
audiences.remove("newsletter")

print(audiences)
````

### Useful list tools
For now, remember these:

- `len(list)` → number of elements
- `.append(value)` → add one element
- `.remove(value)` → remove by value
- `.pop()` → remove and return an element
- `in` → check membership
- `sorted(list)` → return a sorted copy

````python
campaign_costs = [1200, 500, 1800, 900]

print(len(campaign_costs))
print(1800 in campaign_costs)
print(sorted(campaign_costs))
print(campaign_costs)  # original list is unchanged
````

## 4.2 Dictionaries
A **dictionary** stores data as **key-value pairs**.

```python
{ "key": value }
```

Dictionaries are useful when each value has a meaningful label.

````python
campaign = {
    "name": "Summer Launch",
    "budget": 5000,
    "channel": "Social"
}

print(campaign)
````

### Accessing values
Instead of using a position, use the **key**.

````python
campaign = {
    "name": "Summer Launch",
    "budget": 5000,
    "channel": "Social"
}

print(campaign["name"])
print(campaign["channel"])
````

### Adding and modifying values

````python
campaign = {"name": "Summer Launch", "budget": 5000}

campaign["country"] = "Spain"  # add
campaign["budget"] = 6500       # modify

print(campaign)
````

### Useful dictionary tools
For now, remember:

- `.keys()` → keys
- `.values()` → values
- `.items()` → key-value pairs
- `.get(key)` → retrieve safely
- `in` → check whether a key exists

````python
campaign = {
    "name": "Summer Launch",
    "budget": 6500,
    "country": "Spain"
}

print(campaign.keys())
print(campaign.get("country"))
print("budget" in campaign)
````

### Lists inside dictionaries
Values in a dictionary can themselves be lists. This is very common in real datasets and APIs.

````python
campaign = {
    "campaign_id": 1042,
    "name": "Summer Launch",
    "channels": ["email", "social", "search"]
}

print(campaign["channels"])
print(campaign["channels"][0])
````

## 4.3 Sets
A **set** is an unordered collection of **unique** values.

Use a set when uniqueness matters more than order.

````python
visitor_sources = {"email", "organic", "email", "paid", "organic"}
print(visitor_sources)
````

Sets keep only unique values. You can add a new element with `.add()`. Adding the same value twice does not create a duplicate.


````python
campaign_tags = set()
campaign_tags.add("summer")
campaign_tags.add("social")
campaign_tags.add("summer")

print(campaign_tags)
````

## 4.4 Tuples
A **tuple** is ordered like a list but **immutable**: once created, its elements cannot be replaced.

Tuples are useful for fixed collections of values.

````python
campaign_period = (2026, "Q3")
print(campaign_period[0])

# Tuples can also store values calculated earlier:
total_leads = 320
conversion_rate = 4.8
campaign_summary = (total_leads, conversion_rate)
print(campaign_summary)
````

## 5. Choosing a data structure
A quick rule of thumb:

| Need | Good choice |
|---|---|
| ordered sequence you will modify | `list` |
| labelled attributes / key-value data | `dict` |
| unique values | `set` |
| fixed ordered values | `tuple` |

## Nested structures
Real data often combines lists and dictionaries. The key strategy is to move **one level at a time** and keep asking: *what type do I have now?*

````python
agency = {
    "name": "Northstar Media",
    "campaigns": [
        {"name": "Summer Launch", "channel": "Social"},
        {"name": "Back to School", "channel": "Search"}
    ]
}

print(type(agency))
print(type(agency["campaigns"]))
print(type(agency["campaigns"][0]))
print(agency["campaigns"][0]["name"])
````

## `==` versus `is` — reference only
- `==` asks whether two objects have the **same value**.
- `is` asks whether they are literally the **same object in memory**.

For the kinds of comparisons you will make in the first weeks, you will usually want `==`.

````python
a = [1, 2]
b = [1, 2]

print(a == b)
print(a is b)
````
