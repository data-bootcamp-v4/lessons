## 1. What can go wrong?

So far, many of our programs assume that users enter exactly what we expect.

Imagine a cinema booking system asking for the number of tickets:


````python
tickets = int(input("How many tickets would you like? "))
````

If the user types `three`, Python cannot convert that string to an integer and raises a `ValueError`.

Errors stop the normal execution of the program unless we handle them.


### Common exceptions you will often meet

You do **not** need to memorize every Python exception. For now, recognize these common ones:

- `ValueError` → a value has the right general type, but cannot be used as expected (for example `int("three")`).
- `TypeError` → an operation receives an incompatible type.
- `KeyError` → a dictionary key does not exist.
- `IndexError` → a sequence index does not exist.

For this lesson, **`ValueError` is the most important one** because user input frequently needs conversion.

## 2. `try` / `except`: keep the program alive

Put the risky operation inside `try`. If the expected exception occurs, Python jumps to the matching `except` block.


````python
try:
    tickets = int(input("How many tickets would you like? "))
    print(f"You selected {tickets} tickets.")
except ValueError:
    print("Please enter the number using digits.")
````

## 3. Catch specific exceptions

Prefer catching the exception you expect:

```python
except ValueError:
```

rather than a bare `except:`. A specific exception makes the intention clearer and avoids accidentally hiding unrelated bugs.


### Reading the exception message with `as`

Sometimes Python's error message is useful. You can store the exception object in a variable:


````python
try:
    row = int("front")
except ValueError as error:
    print(f"Could not convert the row: {error}")
````

## 4. Python errors vs business-rule errors

Some inputs do not cause a Python exception but still make no sense for the program.

For example, `-3` can be converted to an integer successfully, but **negative tickets are not valid**.

We can check the rule ourselves and deliberately raise an exception.


````python
tickets = -3

if tickets < 0:
    raise ValueError("The number of tickets cannot be negative.")
````

## 5. The key pattern: `while` + `try/except`

Handling an error once is useful, but interactive programs usually need to **ask again**.

The pattern is:

1. Start a loop.
2. Ask for input inside `try`.
3. Validate the value.
4. Leave the loop only when the value is valid.
5. If an expected error occurs, explain the problem and repeat.


````python
valid_input = False

while not valid_input:
    try:
        tickets = int(input("How many tickets would you like? "))

        if tickets < 0:
            raise ValueError("Ticket quantity cannot be negative.")

        valid_input = True

    except ValueError as error:
        print(f"Invalid input: {error}")

print(f"Booking {tickets} tickets.")
````

### Why this works

- `while not valid_input` keeps repeating while the input is invalid.
- `int(...)` may raise a `ValueError` for text such as `three`.
- `raise ValueError(...)` handles values that are numeric but violate our rule.
- `valid_input = True` ends the loop only after all checks pass.

## 6. Another useful pattern: validate a choice

Not every invalid input comes from type conversion. Sometimes the user enters a valid string that is not one of the allowed options.


````python
valid_cities = ["Madrid", "Barcelona", "Valencia"]
valid_city = False

while not valid_city:
    try:
        city = input("Choose a destination: ")

        if city not in valid_cities:
            raise ValueError("Destination is not available.")

        valid_city = True

    except ValueError as error:
        print(error)

print(f"Destination selected: {city}")
````

## 7. `else` and `finally` — useful, but not the main pattern

Python also supports:

- `else` → runs when the `try` block finishes **without** an exception.
- `finally` → runs **whether an exception happens or not**.

They are useful, but they are not required for the core retry pattern you will use in the Lab.


````python
try:
    score = int("85")
except ValueError:
    print("Invalid score")
else:
    print("Score converted successfully")
finally:
    print("Validation attempt finished")
````
