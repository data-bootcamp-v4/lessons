### Arrays and lists: one important difference

Multiplying a list by two **repeats** its elements; multiplying a numeric NumPy array by two **multiplies each element**. NumPy arrays normally have one common element type (`dtype`).

````python
numbers = [10, 20, 30, 40, 50]
array = np.array(numbers)

print("List:", numbers * 2)
print("Array:", array * 2)
print("Data type:", array.dtype)
````

### Indexing and slicing — first with Python lists

`values[start:stop]` selects elements from `start` **up to, but not including**, `stop`. Indexing starts at zero. Leaving out `start` means “from the beginning”; leaving out `stop` means “to the end”.

**Important:** We did not cover slicing when studying lists, so this is a new concept.

````python
values = [10, 20, 30, 40, 50]

print("Python list:")
print(values[1])       # One element
print(values[1:4])     # Positions 1, 2 and 3
print(values[:3])      # First three elements
print(values[2:])      # From position 2 onwards

# Apply exactly the same slicing to a NumPy array
array_values = np.array(values)

print("\nNumPy array:")
print(array_values[1])
print(array_values[1:4])
print(array_values[:3])
print(array_values[2:])
````

### Two-dimensional arrays: rows and columns

A 2D array has two positional axes. Use `[row, column]` to select one value and `[row_slice, column_slice]` to select a region.

````python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Shape:", matrix.shape)
print("Second row, third column:", matrix[1, 2])
print("First two rows, last two columns:\n", matrix[0:2, 1:3])
print("Entire second column:", matrix[:, 1])
````

## 2. Why Pandas?

NumPy arrays are useful for numerical data, but business datasets usually need **row and column labels** and often contain **different types of variables**.

Pandas provides:

- **Series:** one-dimensional labelled data, similar *in use* to a dictionary with labels and values;
- **DataFrame:** a two-dimensional table whose columns can be thought of as Series sharing a row index.

This is a conceptual analogy, not a claim about how Pandas stores every value internally.

## 3. Pandas Series — labelled one-dimensional data

A **Series** associates each value with an **index label**. A useful first analogy is a Python dictionary: **labels → values**. A Series is not a dictionary, but the way we retrieve values by label feels familiar.

We will first create Series independently, and then combine several Series into a DataFrame.

### From a Python dictionary to a Series

````python
city_temperatures = {"Madrid": 18, "Paris": 21, "Berlin": 19}

temperatures = pd.Series(city_temperatures)

print(temperatures)
print("Temperature in Paris:", temperatures["Paris"])
````

### From a list: default and custom indexes

When no index is provided, Pandas creates integer labels starting at zero. We can also supply meaningful labels explicitly.

````python
delivery_times = pd.Series([42, 28, 55], name="delivery_minutes")
print(delivery_times)

labelled_times = pd.Series([42, 28, 55], index=["D101", "D102", "D103"])
print(labelled_times)
````

### A few useful Series operations

Like NumPy arrays, numeric Series support element-wise calculations. Series also offer statistical methods and methods for counting categories.

````python
print("Mean time:", delivery_times.mean())
print("Median time:", delivery_times.median())

cities = pd.Series(["Madrid", "Barcelona", "Madrid", "Valencia", "Barcelona", "Madrid"])
print(cities.value_counts())
````

## 4. Pandas DataFrames — combining labelled columns

A **DataFrame** is a two-dimensional table whose columns can be understood as Series sharing a common row index. This is a useful conceptual model, although Pandas handles alignment and data storage internally.

First, combine a few Series with the **same index** into a DataFrame.

````python
ids = pd.Series([101, 102, 103], index=["A", "B", "C"])
cities_small = pd.Series(["Madrid", "Barcelona", "Valencia"], index=["A", "B", "C"])
minutes = pd.Series([42, 28, 31], index=["A", "B", "C"])

small_deliveries = pd.DataFrame({
    "delivery_id": ids,
    "city": cities_small,
    "delivery_minutes": minutes
})
small_deliveries
````

Notice that the three columns share the row labels `A`, `B`, and `C`. Pandas uses these labels to align the Series when constructing the DataFrame.

For convenience, we can also create a DataFrame directly from a dictionary of lists. We will use this slightly larger **logistics** dataset for the rest of the lesson.

````python
deliveries = pd.DataFrame({
    "delivery_id": [101, 102, 103, 104, 105, 106],
    "city": ["Madrid", "Barcelona", "Madrid", "Valencia", "Barcelona", "Madrid"],
    "vehicle": ["Van", "Bike", "Van", "Bike", "Van", "Bike"],
    "distance_km": [18.5, 6.2, 24.0, 8.1, 15.3, 4.8],
    "delivery_minutes": [42, 28, 55, 31, 39, 22],
    "delivered": [True, True, False, True, True, True]
})

deliveries
````

## 7. Understanding categorical variables

For categorical variables, ask: **Which categories exist? How many are there? How frequent is each one?**

- `unique()` lists distinct values;
- `nunique()` counts them;
- `value_counts()` counts observations per category;
- `describe(include="object")` provides a summary for text columns (count, unique, top, freq).

````python
print("Unique cities:")
print(deliveries["city"].unique())

print("\nNumber of unique cities:")
print(deliveries["city"].nunique())

print("\nFrequency of each city:")
display(deliveries["city"].value_counts())
````

Because `value_counts()` returns a Series, we can sort it.

````python
deliveries["city"].value_counts().sort_values(ascending=True)
````

````python
print("Categorical summary:")
display(deliveries.describe(include="object"))

city_counts = deliveries["city"].value_counts()
print("Most frequent city:", city_counts.idxmax())
````

## 9. Selecting data with `loc` and `iloc`

**Connection with NumPy:** the syntax of `matrix[1:3, 0:2]` is almost identical to `deliveries.iloc[1:3, 0:2]`. Both select by **position** and exclude the stop position.

- `.iloc[row_positions, column_positions]` uses integer positions;
- `.loc[row_labels, column_labels]` uses labels.

**Important difference:** when slicing labels with `.loc`, the final label is normally **included**.

With the default integer index, labels and positions may look the same. They need not be.

````python
print("NumPy positional slice:")
print(matrix[0:2, 1:3])

print("Pandas positional slice:")
display(deliveries.iloc[0:2, 1:3])
````

````python
display(deliveries)

# Using labels: row label 1, column label "city"
print(deliveries.loc[1, "city"])

# Using positions: second row, second column
print(deliveries.iloc[1, 1])
````

### When labels and positions are different

To make the distinction visible, let's give the rows custom labels.

````python
labelled_deliveries = deliveries.copy()
labelled_deliveries.index = ["A", "B", "C", "D", "E", "F"]

display(labelled_deliveries)
````

Now the second row has:

- label → `"B"`
- position → `1`

Both expressions below select the same cell, but they do so in different ways.

````python
print(labelled_deliveries.loc["B", "city"])   # labels
print(labelled_deliveries.iloc[1, 1])        # positions
````

## 11. Basic numerical summaries

For a numerical Series, useful methods include `.mean()`, `.median()`, `.mode()`, `.std()`, `.min()`, `.max()` and `.quantile()`.

an unfamiliar dataset will ask you to interpret these results, not just calculate them.

````python
print("Mean:", deliveries["delivery_minutes"].mean())
print("Median:", deliveries["delivery_minutes"].median())
print("75th percentile:", deliveries["delivery_minutes"].quantile(0.75))
````

### Filtering using a calculated threshold

Define a long delivery as one above the 75th percentile.

````python
threshold = deliveries["delivery_minutes"].quantile(0.75)
long_deliveries = deliveries.loc[deliveries["delivery_minutes"] > threshold]
long_deliveries
````
