## 1. Why Pandas?

Python includes useful built-in data structures such as lists, dictionaries, tuples and sets.

However, native Python does **not** include a data structure specifically designed for working with tabular data such as:

- CSV files;
- spreadsheets;
- database tables.

For this kind of work, we normally use an external library such as **Pandas**.

Pandas provides two main tabular data structures:

- **Series** → one-dimensional labelled data;
- **DataFrame** → two-dimensional labelled data.

### Standard library vs external libraries

Python includes a **standard library** with modules that are available as soon as Python is installed, for example:

```python
import math
import datetime
```

Other libraries, such as Pandas, NumPy or Matplotlib, are **external libraries** and normally need to be installed separately.

In this bootcamp you are using **Anaconda**, which already includes Pandas and many other common data-analysis libraries, so you do not need to install Pandas manually.

### Importing libraries

Before using a library, we need to import it into our Python program.

There are three common patterns:

```python
import library
import library as alias
from library import something
```

For Pandas, the standard convention is:

```python
import pandas as pd
```

The alias `pd` simply gives us a shorter name to use when calling Pandas functions and classes.

````python
import pandas as pd
````

## 7. Selecting data with `loc` and `iloc`

Rows and columns in a DataFrame have **labels**.

They also have **positions**:

- first row → position `0`
- second row → position `1`
- first column → position `0`
- second column → position `1`

Pandas gives us two main ways to select data:

- `loc` → select using **labels**
- `iloc` → select using **integer positions**

With the default DataFrame index, row labels are `0, 1, 2, ...`, so labels and positions can *look* identical. They do not have to be the same.

````python
display(deliveries)

# Using labels: row label 1, column label "city"
print(deliveries.loc[1, "city"])

# Using positions: second row, second column
print(deliveries.iloc[1, 1])
````

### When labels and positions are different

To make the distinction visible, let's give the rows custom labels.

````python
labelled_deliveries = deliveries.copy()
labelled_deliveries.index = ["A", "B", "C", "D", "E", "F"]

display(labelled_deliveries)
````

Now the second row has:

- label → `"B"`
- position → `1`

Both expressions below select the same cell, but they do so in different ways.

````python
print(labelled_deliveries.loc["B", "city"])   # labels
print(labelled_deliveries.iloc[1, 1])        # positions
````

## 9. Basic numerical summaries

Useful Series methods include `mean()`, `median()`, `mode()`, `std()`, `min()`, `max()` and `quantile()`. You do not need to memorize all of them.

````python
print("Mean:", deliveries["delivery_minutes"].mean())
print("Median:", deliveries["delivery_minutes"].median())
print("75th percentile:", deliveries["delivery_minutes"].quantile(0.75))
````

### Filtering using a calculated threshold

Define a long delivery as one above the 75th percentile.

````python
threshold = deliveries["delivery_minutes"].quantile(0.75)
long_deliveries = deliveries.loc[deliveries["delivery_minutes"] > threshold]
long_deliveries
````
